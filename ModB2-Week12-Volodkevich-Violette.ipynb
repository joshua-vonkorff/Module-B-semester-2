# 2. Weekly graph question

Suppose you plot the number visitors to a museum over the seven days of a week; then again the following week.  Can you suggest the advantages and disadvantages of the following two ways of presenting the data?

```python
import numpy as np
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
np.random.seed(0)
time_series_0 = np.array([100, 110, 130, 120, 100, 90, 80])
time_series_1 = np.array([70, 90, 100, 100, 120, 110, 90])
fig, axs = plt.subplots(1, 2)
axs[0].plot(time_series_0)
axs[1].plot(time_series_1)
plt.subplots_adjust(wspace = 0)
plt.xlabel("Day of the week")
plt.ylabel("Number of visitors")
```

```python
plt.plot(time_series_0)
plt.plot(time_series_1)
plt.xlabel("Day of the week")
plt.ylabel("Number of visitors")
```

Option 1: Side-by-Side Bar Charts
Each day of the week has two bars next to each other—one for each week.

✅ Advantages:
Easy comparison between the same day across two weeks (e.g., Monday vs. Monday).

Clear visual separation between weeks.

Viewers can quickly identify trends or changes for specific days.

⚠️ Disadvantages:
With many categories (e.g., months or different locations), the chart can become cluttered or hard to read.

Small differences in bar height may be hard to detect, especially if the y-axis range is large.

Doesn't show the flow of the week as clearly.

📈 Option 2: Line Charts with Two Lines (One for Each Week)
One line for Week 1 and one for Week 2, plotted across the days of the week.

✅ Advantages:
Highlights patterns and trends across the week (e.g., increasing visits from Monday to Sunday).

Easier to spot weekly rhythms, peaks, or dips in attendance.

Cleaner and easier to interpret when dealing with more than two weeks.

⚠️ Disadvantages:
Precise day-to-day differences between the two weeks are less obvious than with bars.

Viewers may need to look more closely to interpret specific values.

🔍 Summary:
Use side-by-side bars if your goal is to compare the same day across two weeks clearly.

Use line charts if you want to emphasize overall weekly patterns or trends.

# 4. Storytelling With Data plot

Reproduce any graph of your choice in p. 187-205 of the Storytelling With Data book as best you can. (Chapter Eight). You do not have to get the exact data values right, just the overall look and feel.

```python
import matplotlib.pyplot as plt

# Simulated data
departments = ['Marketing', 'Sales', 'R&D', 'Customer Support', 'HR']
spending = [45, 52, 78, 38, 33]

# Highlight R&D as the focal point
highlight_index = departments.index('R&D')

# Plot setup
plt.figure(figsize=(8, 5))
bars = plt.bar(departments, spending, color='gray')

# Emphasize the R&D bar
bars[highlight_index].set_color('#1f77b4')

# Add annotations
for i, bar in enumerate(bars):
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width() / 2, height + 2, f'{height}M', 
             ha='center', va='bottom', fontsize=10)

# Add annotation for key insight
plt.text(highlight_index, spending[highlight_index] + 10, 
         'R&D receives the highest budget\nat $78M in 2024', 
         ha='center', fontsize=10, color='#1f77b4')

# Clean up chart
plt.title('Annual Department Budget Allocation (2024)', fontsize=14)
plt.ylabel('Budget (in Millions)')
plt.xticks(rotation=15)
plt.ylim(0, 100)
plt.grid(axis='y', linestyle='--', alpha=0.4)
plt.tight_layout()

# Show plot
plt.show()
```
